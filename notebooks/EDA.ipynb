# Exploratory data analysis
Descriptive plots use the regenerated snapshot. Do not derive model choices or fitted preprocessing from the full dataset. Training, validation and test groups are created separately by `training/train.py`.

```python
from pathlib import Path
import sys
ROOT = Path.cwd() if (Path.cwd() / "Backend").exists() else Path.cwd().parent
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from training.data import export_snapshots, feature_frame
raw, cleaned = export_snapshots()
features = feature_frame(cleaned)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].hist(cleaned.Fiyat / 1e6, bins=40)
axes[0].set(xlabel="Price (million TRY)", ylabel="Records")
axes[1].hist(np.log1p(cleaned.Fiyat), bins=40)
axes[1].set(xlabel="log1p(price)", ylabel="Records")
fig.tight_layout()
plt.show()
```

```python
counts = cleaned["Şehir"].value_counts()
print(counts.to_string())
print("Cities with fewer than 30 eligible records:", int((counts < 30).sum()))
```

```python
numeric = features.select_dtypes(include="number").copy()
numeric["log_price"] = np.log1p(cleaned.Fiyat)
print(numeric.corr().round(3))
```

Coverage is highly uneven. Small city samples cannot support strong location-specific claims. The original dataset source and collection date are not recorded; figures describe this file only.
